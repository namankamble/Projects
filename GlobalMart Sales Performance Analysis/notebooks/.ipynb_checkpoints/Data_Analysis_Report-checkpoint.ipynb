### ***Import Libraries for EDA:***

```python

import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
import plotly.express as px 
import plotly.graph_objects as go

import warnings
warnings.filterwarnings('ignore')
```


### ***Data Collection:***


```python

data = pd.read_csv('../Data/GlobalMart_Sales_Performance_Analysis.csv',)
data = data.drop(columns=['CustomerID','CustomerEmail'])
data
```


### ***Data Cleaning and Preprocessing:***


```python

data.info()
```

```python

data.isnull().sum().sum()
```

```python

data.duplicated().sum()
```

```python

data.describe()
```

```python

# Data Type Conversion
data['OrderDate'] = pd.to_datetime(data['OrderDate'])
data['ShipDate'] = pd.to_datetime(data['ShipDate'])
```

```python

data.info()
```


### ***Exploratory Data Analysis (EDA)***


```python

print(f"Categorical Collumns: \n{data.select_dtypes(include='object').columns}")
print()
print(f"Categorical Collumns: \n{data.select_dtypes(include='number').columns}")
      
```

***
#### ***Executive Summary:***         


***1. Total Sales and Total Profit***

```python

Total_Sales = sum(data['Sales'])
Total_Profit = sum(data['Profit'])
Profit_Margin = (Total_Profit / Total_Sales) * 100
```

```python

summary_df = pd.DataFrame({
    'Metric': ['Total Sales', 'Total Profit'],
    'Amount': [Total_Sales, Total_Profit]
})

custom_colors = {'Total Sales': '#00BFA6', 'Total Profit': '#FFA500'}

fig = px.bar( summary_df, x='Metric', y='Amount', color='Metric', text=summary_df['Amount'].apply(lambda x: f"{x:,.2f}"), 
             color_discrete_map=custom_colors, title="Total Sales and Profit Overview"
)

fig.update_layout( xaxis_title='Metrics', yaxis_title='Amount', template='plotly_dark', xaxis=dict(showgrid=False), 
                  yaxis=dict(showgrid=True), title_font_size=20, height=500
)

fig.update_traces(
    textposition='outside',
    marker=dict(line=dict(color='white', width=1.5))
)

fig.show()
```

***2. Profit Margin Gauge***

```python

fig = go.Figure(go.Indicator(
    mode="gauge+number",
    value=Profit_Margin,
    title={'text': "Overall Profit Margin (%)"},
    gauge={'axis': {'range': [None, 100]}}
))
fig.update_layout(template="plotly_dark")
fig.show()
```

***3. Trend Over Time (Month/Year)***

```python

data["MonthYearDate"] = (data["OrderDate"].dt.month_name() + "-" + data["OrderDate"].dt.year.astype(str))

monthly_summary = data.groupby('MonthYearDate').agg({
    'Sales':'sum',
    'Profit':'sum'
})

monthly_summary['Profit_Margin'] = (monthly_summary['Profit'] / monthly_summary['Sales']) * 100
```

```python

fig = px.line(monthly_summary, x=monthly_summary.index, y=['Sales', 'Profit'], title='Monthly Sales & Profit Trend',
             color_discrete_map=custom_colors)
fig.update_layout(template='plotly_dark', xaxis_title='Month', yaxis_title='Amount')
fig.show()
```

***

#### ***Sales & Profit Breakdown:***


***1. Sales and Profit by Region.***


```python

by_region = data.groupby('Region').agg({
    'Sales': 'sum',
    'Profit': 'sum'
}).reset_index()

by_region_melted = by_region.melt(id_vars='Region', var_name='Metric', value_name='Amount')

fig = px.bar( by_region_melted, x='Region', y='Amount', color='Metric', barmode='group', 
             text=by_region_melted['Amount'].apply(lambda x: f"{x:,.2f}"), title='Sales and Profit by Region',
            color_discrete_map=custom_colors)

fig.update_layout( xaxis_title='Region', yaxis_title='Amount', template='plotly_dark', height=500, title_font_size=22,
    legend_title_text='Metric', xaxis=dict(showgrid=False), yaxis=dict(showgrid=True))

fig.update_traces(
    textposition='outside',
    marker=dict(line=dict(color='white', width=1.3))
)

fig.show()
```

***2. Sales and Profit by ProductCategory.*** 


```python

by_product_category = data.groupby('ProductCategory').agg({
    'Sales': 'sum',
    'Profit': 'sum'
}).reset_index()

by_product_category_melted = by_product_category.melt(id_vars='ProductCategory', var_name='Metric', value_name='Amount')

fig = px.bar( by_product_category_melted, x='ProductCategory', y='Amount', color='Metric', barmode='group', 
             text=by_product_category_melted['Amount'].apply(lambda x: f"{x:,.2f}"), title='Sales and Profit by Product Category',
             color_discrete_map=custom_colors)

fig.update_layout( xaxis_title='Product Category', yaxis_title='Amount', template='plotly_dark', height=500, title_font_size=22,
    legend_title_text='Metric', xaxis=dict(showgrid=False), yaxis=dict(showgrid=True))

fig.update_traces(
    textposition='outside',
    marker=dict(line=dict(color='white', width=1.3))
)

fig.show()
```

***3. Sales and Profit by ProductName within each category.***


```python

by_productcategory_productname = data.groupby(['ProductCategory', 'ProductName']).agg({
    'Sales': 'sum',
    'Profit': 'sum'
}).reset_index()

by_productcategory_productname_melted = by_productcategory_productname.melt( id_vars=['ProductCategory', 'ProductName'], 
                                                                            var_name='Metric', value_name='Amount')

by_productcategory_productname_melted.sort_values(by='Amount', ascending=False, inplace=True)

fig = px.bar( by_productcategory_productname_melted, x='ProductName', y='Amount', color='Metric',
             title="Sales and Profit by Product within Each Category", color_discrete_map=custom_colors)

fig.update_layout( xaxis_title='Product', yaxis_title='Amount', template='plotly_dark', 
                  title_font_size=20, height=600, showlegend=True )

fig.update_traces(
    textposition='outside',
    marker=dict(line=dict(color='white', width=1.2))
)

fig.show()
```

***

#### ***Customer Insights:***


***1. Sales and Profit contribution by CustomerSegment.***

```python

by_customer_segment = data.groupby('CustomerSegment').agg({
    'Sales': 'sum',
    'Profit': 'sum'
}).reset_index()

melted = by_customer_segment.melt( id_vars='CustomerSegment', var_name='Metric', value_name='Amount' )

melted.sort_values(['Metric', 'Amount'], ascending=[True, False], inplace=True)

fig = px.bar( melted, y='CustomerSegment', x='Amount', color='Metric', orientation='h', barmode='group', 
             text=melted['Amount'].apply(lambda x: f"{x:,.2f}"), title="Sales and Profit Contribution by Customer Segment",
             color_discrete_map=custom_colors)

fig.update_layout( xaxis_title='Amount', yaxis_title='Customer Segment', template='plotly_dark', 
                  title_font_size=20, height=400, legend_title_text='Metric')

fig.update_traces(
    textposition='outside',
    marker=dict(line=dict(color='white', width=1.2))
)

fig.show()
```

***2. Top 10 Customers by Total Sales and Total Profit.***


```python

by_customer = data.groupby('CustomerName').agg({
    'Sales': 'sum',
    'Profit': 'sum'
}).reset_index()

top_customers = by_customer.nlargest(10, 'Sales')

top_customers_melted = top_customers.melt(id_vars='CustomerName', var_name='Metric', value_name='Amount')

fig = px.bar( top_customers_melted, x='Amount', y='CustomerName', color='Metric', orientation='h', barmode='group',
    text=top_customers_melted['Amount'].apply(lambda x: f"{x:,.2f}"), title="Top 10 Customers by Sales (with Profit Comparison)",
    color_discrete_map=custom_colors)

fig.update_layout( xaxis_title='Amount', yaxis_title='Customer Name', template='plotly_dark', title_font_size=20,
    height=500, legend_title_text='Metric'
)

fig.update_traces(
    textposition='outside',
    marker=dict(line=dict(color='white', width=1.2))
)

fig.show()
```

***3. unique Customers and Orders.***

```python

summary_df = pd.DataFrame({
    'Metric': ['Unique Customers', 'Unique Orders'],
    'Count': [data['CustomerName'].nunique(), data['OrderID'].nunique()]
})

fig = px.bar(summary_df, y='Metric', x='Count', text='Count', color='Metric', 
             color_discrete_map=custom_colors,
             title='Unique Customers and Orders')

fig.update_layout( xaxis_title='', yaxis_title='Count', template='plotly_dark', title_font_size=20,
            height=400, showlegend=False )

fig.update_traces(
    textposition='outside',
    marker=dict(line=dict(color='white', width=1.2))
)

fig.show()
```
